```python
import os
import pandas as pd

# Define paths (Update these paths to match your folder structure in Google Drive)
excel_path = '/content/drive/MyDrive/your_file.xlsx'
output_md_path = '/content/drive/MyDrive/gemini_skills.md'

if os.path.exists(excel_path):
    # Load the Excel file
    # By default, this loads the first sheet. You can specify sheet_name='Sheet1'
    df = pd.read_excel(excel_path)

    # Convert DataFrame to Markdown string
    markdown_table = df.to_markdown(index=False)

    # Save to .md file
    with open(output_md_path, 'w', encoding='utf-8') as f:
        f.write("# Gemini Skills Data\n\n")
        f.write(markdown_table)

    print(f"Successfully converted Excel to Markdown and saved to: {output_md_path}")

    # Display preview
    display(df.head())
else:
    print(f"Error: The file at '{excel_path}' was not found. Please verify the file path in your Google Drive.")
```
